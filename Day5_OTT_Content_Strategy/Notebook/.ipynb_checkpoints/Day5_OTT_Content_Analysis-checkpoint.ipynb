```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

BASE_DIR = Path(
    r"D:\Veda Internship\Sakshi\Super_Store_DataCleaning"
    r"\OTT_Streaming_Content_Trend_Analysis_Day31"
    r"\Day5_OTT_Content_Strategy"
)

DATA_PATH = BASE_DIR / "Dataset" / "netflix_titles.csv"
CHART_DIR = BASE_DIR / "Charts"
REPORT_DIR = BASE_DIR / "Reports"


for folder in [CHART_DIR, REPORT_DIR]:
    folder.mkdir(parents=True, exist_ok=True)

df = pd.read_csv(DATA_PATH)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

display(df.head())
```

```python
print("========== FINAL OTT ANALYSIS ==========")

print("\nTotal Titles:", len(df))

print("\nContent Type Distribution:")
print(df["type"].value_counts())

print("\nTop 10 Genres:")
genres = df["listed_in"].dropna().str.split(", ").explode()
print(genres.value_counts().head(10))

print("\nTop 10 Countries:")
countries = (
    df["country"].fillna("Unknown")
    .str.split(",").str[0].str.strip()
)
print(countries[countries != "Unknown"].value_counts().head(10))

print("\nContent Ratings:")
print(df["rating"].fillna("Unknown").value_counts().head(10))
```

```python
movie_duration = pd.to_numeric(
    df.loc[df["type"] == "Movie", "duration"]
      .str.extract(r"(\d+)")[0],
    errors="coerce"
).dropna()

print("Average Movie Duration:", round(movie_duration.mean(), 2))
print("Median Movie Duration:", movie_duration.median())
print("Minimum Duration:", movie_duration.min())
print("Maximum Duration:", movie_duration.max())
```
